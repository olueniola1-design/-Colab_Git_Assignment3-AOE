```python
# Task 1: Load the Titanic dataset
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"
df = pd.read_csv(url)
print("Dataset loaded successfully. Shape:", df.shape)
```

```python
# Task 2: Inspect the data
print("--- First 5 Rows ---")
display(df.head())

print("\n--- Last 5 Rows ---")
display(df.tail())

print("\n--- Data Types ---")
display(df.dtypes)

print("\n--- Summary Statistics ---")
display(df.describe())
```

```python
# Task 3: Clean and preprocess the data

# Create new categorical feature 'PclassCategory' from 'Pclass'
df['PclassCategory'] = df['Pclass'].astype('category')

# Handle missing values in the 'Age' column using mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Create a 'Family' feature by combining 'Siblings/Spouses Aboard' and 'Parents/Children Aboard'
df['Family'] = df['Siblings/Spouses Aboard'] + df['Parents/Children Aboard']

print("Data cleaning and feature engineering completed.")
display(df.head())
```

```python
# Task 4: Identify specific data points
highest_fare_passengers = df[df['Fare'] == df['Fare'].max()]
largest_family_passengers = df[df['Family'] == df['Family'].max()]

print("--- Passengers with the Highest Fare ---")
display(highest_fare_passengers)

print("\n--- Passengers with the Largest Family Size ---")
display(largest_family_passengers)
```

```python
# Task 5: Demonstrate Pandas Series and DataFrame functionalities

# Create a Pandas Series of passenger names
names_series = df['Name']

# Create a DataFrame of passenger names and ages
names_ages_df = df[['Name', 'Age']]

# Slice the DataFrame to show only passengers older than 50
older_passengers = df[df['Age'] > 50]

print("--- Older Passengers (Age > 50) ---")
display(older_passengers.head())
```

```python
# Task 6: Introduce multi-indexing
# Create a multi-index DataFrame of survival rates grouped by 'Sex' and 'Pclass'
multi_index_df = df.groupby(['Sex', 'Pclass'])['Survived'].mean()

print("--- Multi-Index Survival Rates by Sex and Passenger Class ---")
display(multi_index_df)
```

```python
# Task 7: Visualizations

# 1. Histogram of passenger ages
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=30, kde=True, color='skyblue')
plt.title('Histogram of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# 2. Bar plot of survival rates by passenger class
plt.figure(figsize=(8, 5))
sns.barplot(x='Pclass', y='Survived', data=df, palette='viridis', errorbar=None)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()

# 3. Heatmap of correlations between numerical features
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()

# 4. Pair plot of selected features
sns.pairplot(df[['Age', 'Fare', 'Family', 'Survived']], hue='Survived', palette='Set2')
plt.show()
```

```python
# Task 8: Analyze survival rates by sex and passenger class
survival_by_sex = df.groupby('Sex')['Survived'].mean()
print("--- Survival Rates by Sex ---")
display(survival_by_sex)

survival_by_class = df.groupby('Pclass')['Survived'].mean()
print("\n--- Survival Rates by Passenger Class ---")
display(survival_by_class)
```

```python
# Task 9: Analyze age distribution using a box plot
plt.figure(figsize=(12, 6))
sns.boxplot(x='Pclass', y='Age', hue='Sex', data=df, palette='Set3')
plt.title('Age Distribution by Passenger Class and Sex')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
# Task 10: Analyze fare distribution using a violin plot
plt.figure(figsize=(12, 6))
sns.violinplot(x='Pclass', y='Fare', hue='Survived', data=df, split=True, palette='muted')
plt.title('Fare Distribution by Passenger Class and Survival Status')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()

print("EDA tasks completed.")
```
